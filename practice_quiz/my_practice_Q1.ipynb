**Question (C)** Plotting functions with Numpy

<font size = "3">

Using Numpy arrays, create a **line plot** of the function $f(g(t))$, where

$$ f(x) = x^2 - 5x + 2 $$

$$ g(t) = \sin(t + 3) $$

at a set of 75 evenly spaced points in the interval $2\pi \leq t \leq 4\pi$.

```python
# your code here
import numpy as np
t_vals = np.linspace(2*np.pi, 4*np.pi, 75)  

# Create new numpy array with the function evaluated at the same points
g_vals = np.sin(t_vals + 3)
f_of_x = g_vals **2 - 5*g_vals +2

plt.plot(t_vals, f_of_x) # plt.scatter
plt.xlabel("$x$", fontsize = 16)
plt.ylabel("$y$", fontsize = 16)
plt.title("Graph", fontsize = 12)
plt.show()
```
